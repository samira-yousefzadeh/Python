## if/elif

```python
print("Welcome to the rollercoaster!")
```

```python
height = int(input("What is youe height in cm?"))
```

```python
bill = 0
if height>=120:
    print("you can ride the rollercoaster")
    age = int(input("What is your age?"))
    if age<=12:
        print("Child tickets are £3.")
        bill=3
    elif age<=18:
        print("Your tickets are £5.")
        bill=5    
    else:
        print("Adult tickets are £7.")
        bill=7
    wants_photo = input("Would you want a photo take? Type y for Yes and n for No")
    if wants_photo =="y":
        bill+=3
    print(f"Your bill is {bill}")           
else:
    print("Sorry you have to grow taller before you can ride")
```
